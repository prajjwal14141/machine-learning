#### Distribution plots - for quantitative data, can give a specific col or entire dataset at once

1- histogram(binning based, can be of entire dataset or a certain col)  --   sb.histplot(dataset['column'], bins = n, kde = True)....|| n=20 by default

2 - jointplots(two cols) --  sb.jointplot(x= 'col1', y= 'col2',data='dataset', kind = 'scatter) ... || ... kind = scatter/hex/reg/kde ,etc

3- pairplot(all posible pairs of quant. cols)  -- sb.pairplot(data, hue='abc', palette='clr_pattern') ... || ... hue tells for basis of distribution

```python
import seaborn as sb
import numpy as np
import matplotlib.pyplot as plt

df = sb.load_dataset('tips')    #load from github of sb
sb.histplot(df['tip'], bins = 30, kde = 1)
```

```python
# USE PLT.SUBPLOT ALONG WITH FOR MAN PLOTS
plt.subplot(1, 2, 1)
sb.histplot(df['tip'], bins = 30, kde = 1)
plt.subplot(1, 2, 2)
sb.histplot(df['total_bill'])
```

```python
#JOINT-PLOT , PairPlot and rugPlot
sb.rugplot(df['tip']), sb.jointplot(x = 'tip', y='total_bill', data = df), sb.pairplot(df, hue='size', palette='rainbow')
```
