```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
```

### plt functions
.plot(x, y) - draws a line graph

.xlabel("abc") - label for x axis, similar for y

.title("heading") - title for graph


```python
x= np.linspace(0, 5, 11)   # 11 elements with space b/w 0-5
y = x**2

plt.plot(x, y)
plt.title('sqr')
plt.xlabel('abscissa'), plt.ylabel('ordinate')
```
