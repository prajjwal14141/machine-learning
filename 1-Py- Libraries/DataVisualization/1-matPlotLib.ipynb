Matplotlib works on a canvas that can be accesses via plt.figure() fn.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
```

### plt functions
.plot(x, y) - draws a line graph

.xlabel("abc") - label for x axis, similar for y

.title("heading") - title for graph

##### subplots
divide the canvas into several subpolots . use .subplot to create and then plt.plot(...) to plot the specific figure

.subplot(a, b, x) - create a*b subplots of canvas and work on xth subplot


```python
x= np.linspace(0, 5, 11)   # 11 elements with space b/w 0-5
y = x**2

plt.plot(x, y)
plt.title('sqr')
plt.xlabel('abscissa'), plt.ylabel('ordinate')
```

```python
plt.subplot(3, 2, 4)
plt.plot(x, x*8)
plt.plot(x, 3.14*x*x)
plt.subplot(3, 2, 1)
plt.plot(x, y)
```

#### Object Oriented Way
MatPlotLib allows the aspect ratio, DPI and figure size to be specified when the figure is created. You can use the figsize and dpi keyword arguments.

figsize is a tuple of width and height of the figure in inches dpi is the dots-per-inch(pixel per inch).

fig = plt.figure() -- canvas figures

fig = plt.figure(figsize = (a, b), dpi = "200") --    create (a * b) canvas, dpi = 200(default is 100)

plt.figure.add_axes( l, b, w, h )   //  l = distance from left, b = dist from bottom ///  w = width, h = height

```python
fig = plt.figure(figsize=(20, 8), dpi = 50)
axis1 = fig.add_axes( [ 0.6, 0.2, 0.75, 0.75 ] )
axis1.plot(x, y)
axis2 = fig.add_axes( [ 0.8, 0.7, .3, .2 ] )
axis2.plot(x, x*2)
```
