```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Name': ['A', 'B', 'C', np.nan],
    'Age' : [1, 2, 3, 1290],
    'City': ['a', 'b', 'c', 's'],
    'Sal.': [100, 200, np.nan, 1290]
}
df = pd.DataFrame(data)
data2 = {
    'A': ['A', 'B', 'C', np.nan, np.nan, np.nan],
    'B' : [1, np.nan, 2, 3, np.nan, 1290],
    'C': ['a', 'b', np.nan, 'c', 's', np.nan],
    'D.': [np.nan, 100, 200, np.nan, 1290, np.nan]
}
df2 = pd.DataFrame(data2)
```

df.isna() -- displays null values, 

df.isna().sum() -- gives number of null values in each row

df.isna().any() -- return columns with null values

```python
x = df.isna()
y = x.sum(), x.any()
x, y
```

##### Removing and filling NaN values
df.dropna(thresh = x) - removes all rows with atleast 'x' NaN values

df.fillna(0) - fills all NaN with 0

df.fillna(value = values) // values is the column values provided 

```python
values = {'A': 120, 'B': 36, 'C': 3, 'D': 23}
df2.fillna(value=values)
```
