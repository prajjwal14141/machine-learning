#### MERGE
pd.merge(df1, df2, on = 'column_name', how = 'inner') 

###### how = 'inner' -- only common rows 
###### how = 'outer' -- all rows 
###### how = 'left' -- include all rows of left dataframe
###### how = 'outer' -- include all rows of right dataframe

```python
import numpy as np
import pandas as pd

employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
pd.merge(employees, salaries, on='employee_id', how='inner'), pd.merge(employees, salaries, on='employee_id', how='outer')
```

```python
pd.merge(employees, salaries, on='employee_id', how='left'), pd.merge(employees, salaries, on='employee_id', how='right')
```
