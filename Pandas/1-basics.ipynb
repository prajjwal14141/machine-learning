### Pandas Series

A 1-D labelled array that can store varying data types. The axis labels are called index. Series elements are vertically stacked

pd.Series(data_holder, index = idx) - creates a series of data in the data_holder, with indices in idx


```python
import pandas as pd
import numpy as np
```

```python
labels = ['a', 'b', 'c']
my_list = [10, 20, 30]
arr = np.array([1, 2, 3])
d = {1:10, 2:20, 3:30}
```

```python
pd.Series(my_list)
```

```python
pd.Series(arr, index = labels)
```

```python
pd.Series(d)
```

## DataFrames

Several series kept together in parallel

###### Creation - pd.DataFrame(data_holder)
###### Indexing - df2['col_name']  , df[['col1', 'col2']]
###### Add new column - df_name['col_name] = [...values...]
###### Delete column  - df_name.drop([col_names], axeis = 1, inplace=True)    // axis =0 to delete row, inplace false for temp 'deletion'
###### Row selection - df.loc[[rowss]]      // iloc for index location
##### Subset Selectin - df.loc[[rows]][[cols]] 
##### conditional selection - df[df["age"] > 30]


```python
#from dictionary 
data = {
    'Name': ['A', 'B', 'C'],
    'Age' : [1, 2, 3],
    'City': ['a', 'b', 'c'],
    'Sal.': [100, 200, 300]
}

pd.DataFrame(data)
```

```python
#from list

col = ['P', 'Q', 'R', 'S']
list1 = [
    ['A', 10, 'abc', 67],
    ['B', 20, 'cde'],
    ['C', 30, 'efg', 3897],
    ['D', 40, 32, 'ijk']
]

df2 = pd.DataFrame(list1)   # no column heading
df2 = pd.DataFrame(list1, columns= col)     #Column headings provided
df2
```

```python
df2['T'] = ["ac", "fgb", "fg", "buh"]        # adding new column
df2['R']                             # display 'R' values
```

```python
df2[['P', 'S', 'T']]
```

```python
df2.loc[[1, 2]][['P', 'S']]
```

```python
df2[(df2['Q'] >=20) &  (df2['P'] >= 'C')]
```
