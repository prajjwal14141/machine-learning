## NumPy Arrays

np.array(1 list as args)

```python
import numpy as np
```

```python
arr_1d = np.array([1,2,3]);    # only 1 list(array) as argument
arr_2d = np.array([ [1,2,3], [4,5,6] ]);

print("1D ARRAY: ", arr_1d);
print("2D ARRAY: ", arr_2d);
```

in list, elements repeated when list is multiplied by scalar;
but in np_array- elements value are in-place doubled, tripled, etc

##### creation
 arange(start, stop, step)- stop is most needed

 zeros((row, col)) - zeros full matrix
 
 ones ((row, col)) - ones full matrix
 
 full ((row, col)) - cosntant val full matrix
 
 random((row, col))   - random matrix

```python
py_list = [1,2,3]
np_array = np.array([1,2,3])

print("List: ", py_list*2)         
print("List: ", np_array*2)
```

```python
zeros = np.zeros((2,3))
ones  = np.ones((4,2))
random = np.random.random((2,3)) 
seq = np.arange(1,17,2)

print(zeros, "\n", ones, "\n", random , "\n", seq)
```

```python
import time
start = time.time()
py_list = [i*2 for i in range (10000000)]
print("List operation took time: ", time.time() - start)

start = time.time()
np_array = np.arange(10000000) * 2  #arange used to create array till this stop
print("Numpy array operation took time: ", time.time() - start)
```
